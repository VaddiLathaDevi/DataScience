```python
import os
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error,mean_absolute_error,mean_absolute_percentage_error
```

```python
os.getcwd()
os.chdir(r"F:\DataScience")
```

```python
df=pd.read_excel("Mtcars_With_Labels_32_Records.Xlsx",sheet_name="mtcars_raw")
```

```python
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.columns
```

```python
y=df['mpg']
```

```python
X=df.drop(['mpg','model'],axis=1)
```

```python
X.head()
```

```python
X.isnull().sum()
```

```python
y.isnull().sum()
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42,shuffle=True)
```

```python
model=LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
y_train_predict=model.predict(X_train)
y_test_predict=model.predict(X_test)
```

```python
print(y_train_predict)
print(y_test_predict)
```

```python

### For test metrics
mse_test=mean_squared_error(y_test,y_test_predict)
rmse_test=np.sqrt(mse_test)
mae_test=mean_absolute_error(y_test,y_test_predict)
mape_test=mean_absolute_percentage_error(y_test,y_test_predict)

print("mse_test:",mse_test)
print("rmse_test:",rmse_test)
print("mae_test:",mae_test)
print("mape_test:",mape_test)
```

```python
## For train metrics
mse_train=mean_squared_error(y_train,y_train_predict)
rmse_train=np.sqrt(mse_train)
mae_train=mean_absolute_error(y_train,y_train_predict)
mape_train=mean_absolute_percentage_error(y_train,y_train_predict)

print("mse_train:",mse_train)
print("rmse_train:",rmse_train)
print("mae_train:",mae_train)
print("mape_train:",mape_train)
```

```python
print(type(model.coef_))     #numpy.ndarray can't directly write in to excel  : so convert int o data frames
print(type(X_train))         #pandas.DataFrame can write directly to array
print(type(X_test))


#m is the coefficient (also called the slope or weight in ML terms).
#y=(m1x1)+(m2x2)+...+c
#m1x1,mx2,... are coefficients
#Coefficient(s): The values multiplying your input features (x). They show how much the output changes when the input changes.
#(c=intercept)
#model_coef_df = pd.DataFrame({'Features':X.columns,'Model Coefficients':model.coef_})
df_model_coefs= pd.DataFrame({'Features':X.columns,'Coefficients:':model.coef_})  
df_intercept=pd.DataFrame({'Intercept:':[model.intercept_]})
df_train_metrics=pd.DataFrame({'Metrics':['mse','rmse','mae','mape'],
                      'Values' :[mse_train,rmse_train,mae_train,mape_train]
                      })
df_train_metrics=pd.DataFrame({'Metrics':['mse','rmse','mae','mape'],
                      'Values' :[mse_train,rmse_train,mae_train,mape_train]
                      })
df_test_metrics=pd.DataFrame({'Metrics':['mse','rmse','mae','mape'],
                      'Values' :[mse_test,rmse_test,mae_test,mape_test]
                      })
print("df_model_coefs:",df_model_coefs)
print("df_intercept:",df_intercept)
print("df_train_metrics:",df_train_metrics)
print("df_test_metrics:",df_test_metrics)
```

```python
#In excel to get like this :X(x1,x2..x10),y(mpg),Predict
#cyl	disp	hp	drat	wt	qsec	vs	am	gear	carb	mpg	Y_Test_Pred
#cyl	disp	hp	drat	wt	qsec	vs	am	gear	carb	mpg	Y_Train_Pred
y_train_pred_df = pd.DataFrame(
    {'Y_Train_Predict': y_train_predict},
    index=y_train.index
)
X_train_final= pd.concat([X_train,y_train,y_train_pred_df],axis=1)


y_test_pred_df=pd.DataFrame(
    {'Y_test_Predict':y_test_predict} ,
    index=y_test.index)

X_test_final=pd.concat([X_test,y_test,y_test_pred_df],axis=1)
```

```python
## Create a Book Sheet for excel
with pd.ExcelWriter("Multiple_regression_19092026.xlsx",engine="xlsxwriter") as writer:
     df.to_excel(writer,sheet_name="RawData",index=True)
     X_train_final.to_excel(writer,sheet_name="TrainingData",index=False)
     X_test_final.to_excel(writer,sheet_name="TestingData",index=True)
     df_model_coefs.to_excel(writer,sheet_name="Coefficients",index=True)
     df_intercept.to_excel(writer,sheet_name="Intercept",index=True)
     df_train_metrics.to_excel(writer,sheet_name="Metrics in Train",index=True)
     df_test_metrics.to_excel(writer,sheet_name="Metrics in Test",index=True)
     
```
