```python
df_number.isnull()
#isnull() -> True if that value is missing, False otherwise
```

```python
#To find how many True is:1 and False is:0 [To Find Doing sum]
df_number.isnull().sum()
```

```python
#How to replace the missing values (mean/median) For contineous variable
#doing on new variable for safer side
#For a continuous variable that is roughly normally distributed (not skewed), missing values are usually replaced with:The mean of the variable
# For a continuous variable that is roughly skewed), median of the variable
df_number['Age_new']=df_number['age'].fillna(df_number['age'].mean())
```
